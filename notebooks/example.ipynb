# Colour map from an x, y, z table
Run with the project's `.venv` kernel.

```python
import sys; sys.path.insert(0, '..')
import numpy as np
from pdm import read_xyz, plot_map, save, MapOptions, Label, LiveMap
```

## Plot the sample file

```python
data = read_xyz('../examples/synthetic_PL.csv')   # or your own x, y, z table
mf = plot_map(data, opts=MapOptions(z_round=20))
save(mf, '../output/notebook_map', formats=['png', 'pdf'])
```

## Live mode (simulated measurement)
Replace the loop body with the real acquisition.

```python
x, ys = data.x, data.y
live = LiveMap(x, y_plan=(ys[0], ys[-1], 0.1), labels=data.labels)
for i, v in enumerate(ys):
    live.add(v, data.z[i])      # spectrum just measured at gate voltage v
live.save('../output/live_demo', formats=['png'])
```
